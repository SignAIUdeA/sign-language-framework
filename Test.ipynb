```python
from data_loading.infrastructure.file_system_repository import FileSystemVideoRepository
from data_loading.usecases.load_videos_use_case import LoadVideosUseCase
from skeleton_extraction.infrastructure.mediapipe_extractor import MediaPipeSkeletonExtractor
from skeleton_extraction.usecases.extract_skeleton_use_case import ExtractSkeletonUseCase
```

```python
data_folder = 'data'
video_repository = FileSystemVideoRepository(data_folder)
load_videos_use_case = LoadVideosUseCase(video_repository)
```

```python
videos = load_videos_use_case.execute()
```

```python
for video in videos:
    print(f"ID: {video.video_id}")
    print(f"Ruta: {video.file_path}")
    print(f"Etiqueta (clase): {video.label}")
    print(f"Info adicional: {video.additional_info}")
    print("-" * 40)
```

```python
if videos:
    video_to_process = videos[0]
    print("Extrayendo skeletons del video:")
    print(f"Ruta: {video_to_process.file_path}")
    extractor_config = {"model_complexity": 1, "min_detection_confidence": 0.5, "min_tracking_confidence": 0.5}
    skeleton_extractor = MediaPipeSkeletonExtractor(config=extractor_config)
    extract_skeleton_use_case = ExtractSkeletonUseCase(skeleton_extractor)
    skeleton_data = extract_skeleton_use_case.execute(video_to_process.file_path)
    print("Datos de Skeleton extraídos:")
    print(skeleton_data)
else:
    print("No se encontraron videos para procesar.")
```
